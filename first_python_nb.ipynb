```python
# dont use plot, map, lib... the code is harder to write and read and make sense of 
import altair as alt
# alt is a name space, the primary object or type to be used inside of altair is a CHART
```

```python
# altair builds on a theory of graphics 
```

```python
# alt.Chart(merged_list)
# without the .mark_bar() you will get an error because you have to tell it what kind of chart you want
```

```python
alt.Chart(merged_list).mark_bar()
# there is still an error here because you need to encode what data you are visualizing (AKA what X and Y)
```

```python
alt.Chart(merged_list).mark_bar().encode(
    x="latimes_make_and_model",
    y="per_100k_hours"
)
```

```python
alt.Chart(merged_list).mark_bar().encode(
    x="per_100k_hours",
    y="latimes_make_and_model"
)
# the code above makes the bar chart horizontal by swapping the X and Y values
```

```python
alt.Chart(merged_list).mark_bar().encode(
    x="per_100k_hours",
    y=alt.Y('latimes_make_and_model').sort('-x')
)
# the - in front of X sorts by X in descending order
```

```python
alt.Chart(merged_list).mark_bar().encode(
    x="per_100k_hours",
    y=alt.Y('latimes_make_and_model').sort('-x')
).properties(title='Helicopter accident rates')
```

### 12.12 - Other Marks
