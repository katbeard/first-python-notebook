## Chapter 7 – Grouping & Aggregating Data 
(Group by)

```python
accident_list.head()
```

```python
accident_list.groupby("latimes_make_and_model")
```

```python
# the groupby object is a special object that allows you to perform operations on groups of data. It is not a dataframe or a series, but it can be used to create new dataframes or series.
# by doing it this way, we get nifty functionalities 
```

```python
accident_list["total_fatalities"].sum()
```

```python
accident_list.groupby("latimes_make_and_model").size()
```

now tally the fatalities by make and model

```python
accident_list.groupby("latimes_make_and_model")["total_fatalities"].sum()
```

```python
# the above code shows fatalities per model, but it does not have to do with the number of flight hours/ crashes
# how many deaths per flight hour is the question we are trying to answer... 
# part of the job of being a reporter, is asking experts what the right questions are to ask the data / how is it measured
# .sum() or .median() are aggregate functions that are performed on a column within a dataframe
```

```python
accident_list.groupby("latimes_make_and_model").size().reset_index()
# you want to use .reset_index() to create a DATA FRAME
# series is ok, but you want the functionality of a data frame 
# resetting the index restores the index column (the 0, 1, 2, 3 column) and gives you more functionality
```

```python
type(accident_list.groupby("latimes_make_and_model").size().reset_index())
```

```python
accident_list.groupby("latimes_make_and_model").size().rename("accidents").reset_index()
```

```python
accident_counts = accident_list.groupby("latimes_make_and_model").size().rename("accidents").reset_index()
```

```python
accident_counts.head()
```

```python
# why would you use groupby over value counts 
# groupby object is much more powerful than value counts
```

```python
accident_row_count = len(accident_counts)
survey_row_count = len(survey)
if survey_row_count != accident_row_count:
    print("Warning: The number of rows in the survey data ({survey_row_count}) does not equal ({accident_row_count})")
else:
    print("The number of rows in both datasets match. You're golden")
```
